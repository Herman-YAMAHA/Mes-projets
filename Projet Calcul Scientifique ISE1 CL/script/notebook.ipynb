```python
import os
import sys

# Détection automatique du chemin du notebook
# Fonctionne quel que soit l'emplacement du dossier sur la machine
chemin_notebook = os.getcwd()

# Si le notebook est dans "code/", on remonte à la racine du projet
if os.path.basename(chemin_notebook) == 'code':
    chemin_projet = os.path.dirname(chemin_notebook)
else:
    chemin_projet = chemin_notebook

# Définition des chemins des sous-dossiers
chemin_data    = os.path.join(chemin_projet, 'data')
chemin_code    = os.path.join(chemin_projet, 'code')
chemin_outputs = os.path.join(chemin_projet, 'outputs')

# Création des dossiers s'ils n'existent pas encore
for dossier in [chemin_data, chemin_code, chemin_outputs]:
    os.makedirs(dossier, exist_ok=True)

# Se placer à la racine du projet
os.chdir(chemin_projet)

# Vérification
print(f"Racine du projet  : {chemin_projet}")
print(f"Dossier data      : {chemin_data}")
print(f"Dossier code      : {chemin_code}")
print(f"Dossier outputs   : {chemin_outputs}")
```
